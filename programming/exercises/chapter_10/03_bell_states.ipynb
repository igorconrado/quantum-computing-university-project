# Estados de Bell (Emaranhamento)

## Objetivo
Criar e medir os 4 estados de Bell - os estados maximamente emaranhados.

## Os 4 Estados de Bell
- |Φ+⟩ = (|00⟩ + |11⟩) / √2
- |Φ-⟩ = (|00⟩ - |11⟩) / √2
- |Ψ+⟩ = (|01⟩ + |10⟩) / √2
- |Ψ-⟩ = (|01⟩ - |10⟩) / √2

## Referências
- Livro: Capítulo 10
- [Qiskit Textbook](https://qiskit.org/textbook/ch-gates/multiple-qubits-entangled-states.html)

```python
# Imports
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
from qiskit_aer import AerSimulator
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_histogram, plot_bloch_multivector
import matplotlib.pyplot as plt
```

## 1. Estado |Φ+⟩ (Bell State Padrão)

|Φ+⟩ = (|00⟩ + |11⟩) / √2

```python
def create_bell_phi_plus() -> QuantumCircuit:
    """
    Cria o estado |Φ+⟩ = (|00⟩ + |11⟩) / √2
    """
    qc = QuantumCircuit(2)
    qc.h(0)       # Superposição no qubit 0
    qc.cx(0, 1)   # Emaranha com qubit 1
    return qc

# Criar e visualizar
qc_phi_plus = create_bell_phi_plus()
print("Circuito |Φ+⟩:")
print(qc_phi_plus.draw())

state = Statevector.from_instruction(qc_phi_plus)
print(f"\nStatevector: {state}")
print("→ Componentes |00⟩ e |11⟩ com amplitudes iguais")
```

## 2. Estado |Φ-⟩

|Φ-⟩ = (|00⟩ - |11⟩) / √2

```python
def create_bell_phi_minus() -> QuantumCircuit:
    """
    Cria o estado |Φ-⟩ = (|00⟩ - |11⟩) / √2
    """
    qc = QuantumCircuit(2)
    qc.h(0)
    qc.cx(0, 1)
    qc.z(0)  # Adiciona fase -1 ao |11⟩
    return qc

# Criar e visualizar
qc_phi_minus = create_bell_phi_minus()
print("Circuito |Φ-⟩:")
print(qc_phi_minus.draw())

state = Statevector.from_instruction(qc_phi_minus)
print(f"\nStatevector: {state}")
print("→ |00⟩ positivo, |11⟩ negativo")
```

## 3. Estado |Ψ+⟩

|Ψ+⟩ = (|01⟩ + |10⟩) / √2

```python
def create_bell_psi_plus() -> QuantumCircuit:
    """
    Cria o estado |Ψ+⟩ = (|01⟩ + |10⟩) / √2
    """
    qc = QuantumCircuit(2)
    qc.h(0)
    qc.cx(0, 1)
    qc.x(0)  # Flipa qubit 0: |00⟩→|10⟩, |11⟩→|01⟩
    return qc

# Criar e visualizar
qc_psi_plus = create_bell_psi_plus()
print("Circuito |Ψ+⟩:")
print(qc_psi_plus.draw())

state = Statevector.from_instruction(qc_psi_plus)
print(f"\nStatevector: {state}")
print("→ Componentes |01⟩ e |10⟩ com amplitudes iguais")
```

## 4. Estado |Ψ-⟩

|Ψ-⟩ = (|01⟩ - |10⟩) / √2

```python
def create_bell_psi_minus() -> QuantumCircuit:
    """
    Cria o estado |Ψ-⟩ = (|01⟩ - |10⟩) / √2
    """
    qc = QuantumCircuit(2)
    qc.h(0)
    qc.cx(0, 1)
    qc.x(0)
    qc.z(0)  # Adiciona fase relativa
    return qc

# Criar e visualizar
qc_psi_minus = create_bell_psi_minus()
print("Circuito |Ψ-⟩:")
print(qc_psi_minus.draw())

state = Statevector.from_instruction(qc_psi_minus)
print(f"\nStatevector: {state}")
print("→ |01⟩ positivo, |10⟩ negativo")
```

## 5. Medir os Estados de Bell

Execute múltiplas vezes e observe as correlações.

```python
def measure_bell_state(circuit: QuantumCircuit, shots: int = 1024) -> dict:
    """
    Mede um estado de Bell.
    """
    qc = circuit.copy()
    qc.measure_all()
    
    simulator = AerSimulator()
    job = simulator.run(qc, shots=shots)
    return job.result().get_counts()

# Medir cada estado de Bell
bell_states = {
    "|Φ+⟩": create_bell_phi_plus(),
    "|Φ-⟩": create_bell_phi_minus(),
    "|Ψ+⟩": create_bell_psi_plus(),
    "|Ψ-⟩": create_bell_psi_minus()
}

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()

for idx, (name, circuit) in enumerate(bell_states.items()):
    counts = measure_bell_state(circuit, shots=1000)
    plot_histogram(counts, ax=axes[idx], title=f"Estado {name}")

plt.tight_layout()
plt.show()
```

## 6. Verificar Emaranhamento

Observe que:
- Nunca obtemos |01⟩ ou |10⟩ para |Φ±⟩
- Nunca obtemos |00⟩ ou |11⟩ para |Ψ±⟩

Isso é a **correlação perfeita** do emaranhamento!

```python
# Verificar correlações - 1000 shots
print("Verificação de Correlações Perfeitas")
print("=" * 50)

for name, circuit in bell_states.items():
    counts = measure_bell_state(circuit, shots=1000)
    print(f"\n{name}:")
    for outcome, count in sorted(counts.items()):
        print(f"  {outcome}: {count} ({count/10:.1f}%)")
    
    # Verificar correlação
    if name in ["|Φ+⟩", "|Φ-⟩"]:
        anti_corr = counts.get("01", 0) + counts.get("10", 0)
        print(f"  → Resultados anti-correlacionados (01/10): {anti_corr} (esperado: 0)")
    else:
        same = counts.get("00", 0) + counts.get("11", 0)
        print(f"  → Resultados iguais (00/11): {same} (esperado: 0)")
```

## 7. Conclusão

Responda:
- O que torna os estados de Bell "especiais"?
- Por que são chamados de "maximamente emaranhados"?
- Se você medir um qubit e obtiver |0⟩, o que você sabe sobre o outro no estado |Φ+⟩?

**Resposta:**

1. **O que torna os estados de Bell "especiais"?**
   - São estados **maximamente emaranhados**: a correlação entre os qubits é perfeita
   - Não podem ser escritos como produto de estados individuais
   - Formam uma base ortonormal para o espaço de 2 qubits

2. **Por que são chamados de "maximamente emaranhados"?**
   - A entropia de emaranhamento é máxima (1 ebit)
   - Cada qubit individual, se visto separadamente, parece completamente aleatório
   - A informação está toda na correlação, não nos qubits individuais

3. **Se você medir um qubit e obtiver |0⟩, o que sabe sobre o outro no estado |Φ+⟩?**
   - Você sabe com **100% de certeza** que o outro também será |0⟩
   - Em |Φ+⟩: sempre (|00⟩ ou |11⟩) → correlação perfeita
   - Esta correlação existe independente da distância entre os qubits!
