# Portas de Um Qubit

## Objetivo
Praticar todas as portas básicas de um qubit e visualizar seus efeitos na Bloch Sphere.

## Portas Abordadas
- **X (NOT):** Rotação de π em torno do eixo X
- **Y:** Rotação de π em torno do eixo Y
- **Z:** Rotação de π em torno do eixo Z
- **H (Hadamard):** Cria superposição
- **S, T:** Portas de fase

## Referências
- Livro: Capítulo 10
- [Qiskit Textbook](https://qiskit.org/textbook/ch-states/single-qubit-gates.html)

```python
# Imports
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector, plot_histogram
import matplotlib.pyplot as plt
import numpy as np
```

## Função Auxiliar para Visualização

```python
def visualize_gate(gate_name: str, circuit: QuantumCircuit):
    """Visualiza o circuito e o estado na Bloch Sphere."""
    print(f"=== Porta {gate_name} ===")
    display(circuit.draw('mpl'))
    
    state = Statevector.from_instruction(circuit)
    display(plot_bloch_multivector(state))
    
    print(f"Statevector: {state}")
    print()
```

## 1. Estado Inicial |0⟩

```python
# Estado Inicial |0⟩
qc = QuantumCircuit(1)
visualize_gate("Estado Inicial |0⟩", qc)
```

## 2. Porta X (NOT)

```python
# Porta X (NOT) - Flipa |0⟩ para |1⟩
qc = QuantumCircuit(1)
qc.x(0)
visualize_gate("X (NOT)", qc)
# Resultado: |0⟩ → |1⟩ (vetor vai do polo norte para o polo sul)
```

## 3. Porta H (Hadamard)

```python
# Porta H (Hadamard) - Cria superposição
qc = QuantumCircuit(1)
qc.h(0)
visualize_gate("H (Hadamard)", qc)
# Resultado: |0⟩ → |+⟩ = (|0⟩ + |1⟩) / √2 (vetor aponta para +X)
```

## 4. Porta Z

```python
# Porta Z após Hadamard
# Z não faz nada visível em |0⟩ ou |1⟩, mas muda fase em superposição
qc = QuantumCircuit(1)
qc.h(0)  # Primeiro criamos superposição
qc.z(0)  # Z muda |+⟩ → |−⟩
visualize_gate("H → Z", qc)
# Resultado: |+⟩ → |−⟩ = (|0⟩ - |1⟩) / √2 (vetor aponta para -X)
```

## 5. Porta Y

```python
# Porta Y em |0⟩
qc = QuantumCircuit(1)
qc.y(0)
visualize_gate("Y", qc)
# Resultado: |0⟩ → i|1⟩ (rotação de π em torno de Y)
```

## 6. Portas S e T (Fase)

```python
# Porta S (rotação de π/2 em Z)
qc_s = QuantumCircuit(1)
qc_s.h(0)
qc_s.s(0)
visualize_gate("H → S", qc_s)
# S adiciona fase de π/2 (vetor aponta para +Y)

# Porta T (rotação de π/4 em Z)
qc_t = QuantumCircuit(1)
qc_t.h(0)
qc_t.t(0)
visualize_gate("H → T", qc_t)
# T adiciona fase de π/4 (entre +X e +Y)

print("Diferença: S = 90°, T = 45° de rotação no equador da Bloch Sphere")
```

## 7. Composição de Portas

Verifique as seguintes identidades:

```python
# Verificação de identidades
from qiskit.quantum_info import Operator

# X·X = I
qc_xx = QuantumCircuit(1)
qc_xx.x(0)
qc_xx.x(0)
print("X·X = I?", np.allclose(Operator(qc_xx).data, np.eye(2)))
print(f"  X·X = \n{Operator(qc_xx).data}")

# H·H = I
qc_hh = QuantumCircuit(1)
qc_hh.h(0)
qc_hh.h(0)
print("\nH·H = I?", np.allclose(Operator(qc_hh).data, np.eye(2)))

# H·X·H = Z
qc_hxh = QuantumCircuit(1)
qc_hxh.h(0)
qc_hxh.x(0)
qc_hxh.h(0)

qc_z = QuantumCircuit(1)
qc_z.z(0)
print("\nH·X·H = Z?", np.allclose(Operator(qc_hxh).data, Operator(qc_z).data))
```

## 8. Exercício: Criar Estado Específico

Use portas para criar o estado |−⟩ = (|0⟩ - |1⟩) / √2

```python
# Criar estado |−⟩ = (|0⟩ - |1⟩) / √2
# Método: H cria |+⟩, depois Z inverte a fase para |−⟩
# Ou: X seguido de H

# Método 1: H → Z
qc_minus = QuantumCircuit(1)
qc_minus.h(0)
qc_minus.z(0)
visualize_gate("|−⟩ via H→Z", qc_minus)

# Método 2: X → H (equivalente)
qc_minus2 = QuantumCircuit(1)
qc_minus2.x(0)
qc_minus2.h(0)
visualize_gate("|−⟩ via X→H", qc_minus2)

# Verificar que são iguais (até fase global)
s1 = Statevector.from_instruction(qc_minus)
s2 = Statevector.from_instruction(qc_minus2)
print(f"Estados equivalentes? {np.allclose(np.abs(s1.data), np.abs(s2.data))}")
```

## 9. Conclusão

Responda:
- Qual porta cria superposição?
- O que a porta Z faz na prática?
- Por que as portas são chamadas de "rotações"?

**Resposta:**

1. **Qual porta cria superposição?**
   - A porta **Hadamard (H)** cria superposição. Ela transforma |0⟩ em |+⟩ = (|0⟩ + |1⟩)/√2.

2. **O que a porta Z faz na prática?**
   - A porta Z adiciona uma fase de π ao estado |1⟩. Em |0⟩ ou |1⟩ puros, não há efeito visível.
   - Em superposição, Z troca |+⟩ ↔ |−⟩ (inverte o sinal da componente |1⟩).

3. **Por que as portas são chamadas de "rotações"?**
   - Na Bloch Sphere, cada porta corresponde a uma rotação:
     - X: rotação de 180° em torno do eixo X
     - Y: rotação de 180° em torno do eixo Y
     - Z: rotação de 180° em torno do eixo Z
     - H: rotação de 180° em torno do eixo (X+Z)/√2
   - Portas parametrizadas (Rx, Ry, Rz) permitem rotações de ângulos arbitrários.
