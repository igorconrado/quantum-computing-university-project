## 1. Carregar Dados de Teste

```python
def load_cities(dataset: str) -> tuple:
    with open('../data/cities.json', 'r') as f:
        data = json.load(f)
    cities = data[dataset]['cities']
    coords = [tuple(c) for c in data[dataset]['coordinates']]
    return cities, coords

# List available datasets
with open('../data/cities.json', 'r') as f:
    data = json.load(f)
    
for name, info in data.items():
    n = len(info['cities'])
    perms = math.factorial(n - 1)
    print(f"{name}: {n} cities, {perms:,} permutations")
```

## 2. Visualizar Cidades

```python
def plot_cities(cities: list, coords: list, route: list = None):
    """Plot cities and optionally a route."""
    fig, ax = plt.subplots(figsize=(10, 8))
    
    # Plot cities
    xs = [c[0] for c in coords]
    ys = [c[1] for c in coords]
    ax.scatter(xs, ys, s=100, c='blue', zorder=5)
    
    # Label cities
    for i, city in enumerate(cities):
        ax.annotate(city, (coords[i][0], coords[i][1]), 
                   xytext=(5, 5), textcoords='offset points')
    
    # Plot route if provided
    if route:
        for i in range(len(route)):
            start = coords[route[i]]
            end = coords[route[(i + 1) % len(route)]]
            ax.plot([start[0], end[0]], [start[1], end[1]], 'r-', linewidth=2)
    
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.grid(True, alpha=0.3)
    plt.show()

# TODO: Visualize a dataset
cities, coords = load_cities('medium_8')
plot_cities(cities, coords)
```

## 5. Comparação Gráfica

```python
# TODO: Plot execution time vs number of cities
# Show exponential growth of brute force vs polynomial of SA

def plot_comparison(brute_results: dict, sa_results: dict):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # Time comparison
    if brute_results:
        ns = sorted(brute_results.keys())
        times = [brute_results[n]['time'] for n in ns]
        ax1.plot(ns, times, 'ro-', label='Brute Force O(n!)', linewidth=2)
    
    if sa_results:
        ns = sorted(sa_results.keys())
        times = [sa_results[n]['time'] for n in ns]
        ax1.plot(ns, times, 'bs-', label='Simulated Annealing', linewidth=2)
    
    ax1.set_xlabel('Number of Cities')
    ax1.set_ylabel('Execution Time (seconds)')
    ax1.set_title('Execution Time Comparison')
    ax1.legend()
    ax1.set_yscale('log')
    ax1.grid(True, alpha=0.3)
    
    # Solution quality (only where both have results)
    common_ns = set(brute_results.keys()) & set(sa_results.keys())
    if common_ns:
        ns = sorted(common_ns)
        optimal = [brute_results[n]['distance'] for n in ns]
        sa_dist = [sa_results[n]['distance'] for n in ns]
        gap = [(s - o) / o * 100 for s, o in zip(sa_dist, optimal)]
        
        ax2.bar(ns, gap, color='orange', alpha=0.7)
        ax2.set_xlabel('Number of Cities')
        ax2.set_ylabel('Gap from Optimal (%)')
        ax2.set_title('Solution Quality (SA vs Optimal)')
        ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()

# TODO: Call with your results
# plot_comparison(brute_force_results, sa_results)
```

## 6. Projeção de Tempo para n Cidades

```python
# Theoretical time projection for brute force
# Assuming 1 million permutations per second

perms_per_second = 1_000_000

print("Projected Brute Force Execution Time:")
print("=" * 50)
print(f"{'Cities':<10} {'Permutations':<20} {'Time'}")
print("-" * 50)

for n in range(5, 21):
    perms = math.factorial(n - 1)
    seconds = perms / perms_per_second
    
    if seconds < 60:
        time_str = f"{seconds:.2f} seconds"
    elif seconds < 3600:
        time_str = f"{seconds/60:.2f} minutes"
    elif seconds < 86400:
        time_str = f"{seconds/3600:.2f} hours"
    elif seconds < 31536000:
        time_str = f"{seconds/86400:.2f} days"
    else:
        time_str = f"{seconds/31536000:.2f} years"
    
    print(f"{n:<10} {perms:<20,} {time_str}")
```
